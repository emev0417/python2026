**1.1 Descripción general del problema planteado**

En el contexto actual, la demanda de energía crece de forma constante, lo que satura las redes de distribución. Al construir un modelo predictivo basado en regresión (Lineal, Ridge y Lasso), la empresa puede anticipar picos de consumo basándose en factores meteorológicos externos (temperatura, viento, humedad exterior) y condiciones internas de la vivienda. **2.1 Análisis del Conjunto de Datos**
* ¿Cuántos registros posee el dataset?:El conjunto de datos posee 19,735 registros (filas). Cada registro representa una medición tomada en intervalos de 10 minutos durante un periodo de aproximadamente 4 meses y medio.

* ¿Cuántas variables contiene?:Contiene 29 variables (columnas) en total.

* ¿Qué tipos de datos existen?:

    * 1 variable tipo objeto / string (date):Que registra el año-mes-día y la hora de la medición.
    * 26 variables de tipo flotante (float64): Que representan las mediciones continuas de temperatura (T1 a T9), humedad (RH_1 a RH_9) y las variables meteorológicas de la estación del clima.
    * 2 variables de tipo entero (int64): Que corresponden a la variable objetivo Appliances y a la variable predictora lights.

* ¿Cuál será la variable objetivo del modelo?: La variable objetivo (target) es Appliances. Es una variable numérica continua que representa el consumo de energía de los electrodomésticos de la vivienda medido en vatios-hora (Wh). Al ser continua usaremos algoritmos de regresión.

```python
import pandas as pd

#Cargar el dataset desde la fuente oficial de UCI
url = "https://archive.ics.uci.edu/static/public/374/appliances+energy+prediction.zip"
df = pd.read_csv(url, compression='zip')

#Obtener dimensiones (Registros y Variables)
filas, columnas = df.shape

print("ESTRUCTURA DEL DATASET")
print(f"¿Cuántos registros posee el dataset?: {filas} filas")
print(f"¿Cuántas variables contiene?: {columnas} columnas")
print("\nTIPOS DE DATOS POR COLUMNA")
print(df.dtypes)

print("\nRESUMEN GLOBAL DE TIPOS")
print(df.dtypes.value_counts())
```

* ¿Qué variables podrían influir en el consumo energético?

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Configuración del estilo de los gráficos
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (12, 8)

# Calcular la correlación de todas las numéricas con la variable objetivo
# Excluimos 'date' porque es string
df_numeric = df.select_dtypes(include=['float64', 'int64'])
correlaciones = df_numeric.corr()['Appliances'].sort_values(ascending=False)

print(" CORRELACIONES DIRECTAS CON APPLIANCES ")
print(correlaciones)

# Graficar las correlaciones para la captura de pantalla
plt.figure(figsize=(10, 8))
sns.barplot(x=correlaciones.values, y=correlaciones.index, palette="coolwarm")
plt.title("Influencia y Correlación de las Variables sobre el Consumo Energético (Appliances)", fontsize=14)
plt.xlabel("Coeficiente de Correlación de Pearson")
plt.ylabel("Variables del Dataset")
plt.tight_layout()
plt.show()
```


* Basándonos en la matriz de correlación y la gráfica generada, las variables que más influyen en el consumo de los electrodomésticos son:

     * lights (Consumo de luces): Es la variable con mayor correlación positiva con Appliances. Tiene todo el sentido lógico: cuando las personas están activas en casa y encienden las luces, también tienden a usar electrodomésticos (televisor, cocina, computadoras).

     * Temperaturas Internas y Externas (T2, T3, T6, T_out): Se observa una correlación positiva moderada. A medida que la temperatura exterior (T_out) o de habitaciones clave como la cocina (T3) aumentan, el uso de ciertos electrodomésticos cambia (por ejemplo, sistemas de ventilación o refrigeración trabajando más fuerte).

     * Humedad Relativa (RH_out, RH_1, RH_2): Muestran una correlación negativa o inversa. Cuando los niveles de humedad exterior son muy altos, el consumo de electrodomésticos tiende a registrar variaciones a la baja, lo que podría estar ligado a dinámicas estacionales o patrones de comportamiento humano (menos actividad en días muy lluviosos/húmedos).

     * Variables meteorológicas secundarias (Windspeed, Press_mm_hg, Visibility): Tienen una correlación muy cercana a cero, lo que sugiere que por sí solas no influyen de forma lineal directa en el consumo de la casa, aunque podrían interactuar con otras variables en modelos más complejos.

## 2.3 Distribución de Variables (Histogramas)

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Histograma de la Variable Objetivo
sns.histplot(df['Appliances'], bins=40, kde=True, ax=axes[0, 0], color='crimson')
axes[0, 0].set_title('Distribución del Consumo (Appliances)')

# Histograma de Luces
sns.histplot(df['lights'], bins=10, kde=False, ax=axes[0, 1], color='gold')
axes[0, 1].set_title('Distribución del Consumo de Luces (lights)')

# Histograma de Temperatura Interna (Cocina T1)
sns.histplot(df['T1'], bins=30, kde=True, ax=axes[1, 0], color='teal')
axes[1, 0].set_title('Distribución de Temperatura Interna (T1)')

# Histograma de Temperatura Externa
sns.histplot(df['T_out'], bins=30, kde=True, ax=axes[1, 1], color='royalblue')
axes[1, 1].set_title('Distribución de Temperatura Externa (T_out)')

plt.tight_layout()
plt.show()
```

### Interpretación de los Resultados:
* Appliances: Muestra una distribución asimétrica positiva (sesgada a la derecha). La mayoría del tiempo el consumo es bajo y constante (la base de la casa), pero tiene una cola muy larga hacia la derecha producida por eventos aislados de alto consumo.

* Temperaturas (T1 y T_out): Presentan una distribución aproximadamente normal. Esto es bueno para  la regresión lineal tradicional, ya que las variables ambientales tienden a oscilar naturalmente alrededor de promedios estacionales bien definidos.

### 2.4 Matriz de Correlación

```python

plt.figure(figsize=(12, 10))

# Calculamos correlación de las variables representativas
corr_matrix = df[variables_clave].corr()

sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Matriz de Correlación de Pearson (Variables Clave)')
plt.show()
```

### Identificación de Relaciones:
* Variables con correlación positiva fuerte: Existe una colinealidad crítica entre las temperaturas internas y externas. Por ejemplo, T2 (sala) y T_out (exterior) tienen una correlación de 0.80. Lo mismo pasa entre T1 y T2 (0.84).

* Variables con correlación negativa fuerte: T_out (temperatura exterior) y RH_out (humedad exterior) muestran una fuerte correlación inversa de -0.78. Físicamente tiene sentido: a mayor calor exterior, la humedad relativa del aire disminuye drásticamente.

* Variables con correlación débil: Las variables como la velocidad del viento (Windspeed) y la presión atmosférica (Press_mm_hg) tienen correlaciones extremadamente cercanas a cero con el consumo (Appliances), oscilando entre -0.01 y 0.06.

* Variable con mayor relación respecto al consumo energético: La variable con mayor impacto directo es lights (0.20), seguida por las temperaturas T2 (0.12) y T6 (0.12).

### 2.5 Visualizaciones Adicionales

#### Gráfico Adicional 1: Boxplot de Consumo por Uso de Luces (Detectando Patrones)

```python

plt.figure(figsize=(12, 5))
sns.boxplot(x='lights', y='Appliances', data=df, palette='Set2')
plt.title('Impacto del Estado de Luces en el Consumo de Electrodomésticos')
plt.xlabel('Consumo de Luces (Wh)')
plt.ylabel('Consumo de Electrodomésticos (Wh)')
plt.yscale('log') # Escala logarítmica para manejar el fuerte sesgo visual de los outliers
plt.show()
```

**Interpretación:**
La escala logarítmica revela que a medida que aumenta el consumo de luces (de 0 a 70 Wh), la base de la caja del boxplot se desplaza hacia arriba. Esto confirma visualmente que el encendido de luces actúa como un indicador de "presencia humana y actividad" dentro de la casa, elevando instantáneamente el consumo promedio de energía.

### Gráfico Adicional 2: Scatterplot Dinámico (Temperatura Externa vs. Humedad Externa)

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='T_out', y='RH_out', data=df.sample(2000, random_state=42), alpha=0.5, color='purple')
plt.title('Relación Física entre Temperatura y Humedad Exterior (Muestra de 2000 registros)')
plt.xlabel('Temperatura Exterior (°C)')
plt.ylabel('Humedad Relativa Exterior (%)')
plt.show()
```

**Interpretación:** Este gráfico de dispersión dibuja una clara pendiente descendente no lineal. La alta densidad de puntos en la esquina superior izquierda demuestra que los climas fríos registrados vinieron acompañados de saturaciones de humedad muy altas, afectando indirectamente las demandas de climatización de la casa.

### Gráfico Adicional 3: Gráfico de Línea Temporal (Patrón de Consumo Diario)

```python
# Convertimos temporalmente a datetime para agrupar por hora del día
df['date'] = pd.to_datetime(df['date'])
consumo_por_hora = df.groupby(df['date'].dt.hour)['Appliances'].mean()

plt.figure(figsize=(10, 5))
plt.plot(consumo_por_hora.index, consumo_por_hora.values, marker='o', color='darkorange', linewidth=2.5)
plt.title('Perfil de Consumo Promedio Residencial por Hora del Día')
plt.xlabel('Hora del Día (0 - 23 hrs)')
plt.ylabel('Consumo Promedio de Electrodomésticos (Wh)')
plt.xticks(range(0, 24))
plt.grid(True, linestyle='--')
plt.show()
```

**Interpretación:** Este gráfico revela un patrón cíclico de comportamiento humano espectacular. El consumo tiene su punto más bajo en la madrugada (de 2:00 a 5:00 AM) debido al descanso. Comienza a subir por la mañana y alcanza su pico máximo absoluto entre las 17:00 y 20:00 (5:00 PM a 8:00 PM), justo cuando las personas regresan del trabajo, encienden electrodomésticos y activan la casa.

Como se determinó en la fase exploratoria, este dataset no contiene valores faltantes (0 nulos) en ninguna de sus celdas.
Entonces la estrategia óptima elegida es mantener el dataset intacto sin aplicar técnicas de imputación (como la sustitución por la media, mediana o algoritmos KNN Imputer). Justificamos matemáticamente esta decisión ya que cualquier intento de rellenar datos inexistentes añadiría ruido o sesgos artificiales a la matriz de covarianza, alterando innecesariamente los coeficientes que calcularán los modelos de Regresión Lineal, Ridge y Lasso.
