#### 0) Cargar una de las imágenes histológicas

```python
# Utilizar la librería skimage.io para leer la imagen 'histo_x.jpg' en formato RGB.
# Normalizar la imagen para que los píxeles se encuentren en el rango [0, 1]
# Visualizar la imagen

from skimage import io
import matplotlib.pyplot as plt

# Leer la imagen histológica en formato RGB
img1 = io.imread('img/histo_1.jpg')

# Normalizar los valores de intensidad al rango [0,1]
img1_norm = img1 / 255.0 # se establece en 255 por la cantidad de pixeles maximos para la foto

# Visualizar imagen original
plt.figure(figsize=(8,8))
plt.imshow(img1_norm)
plt.title('Imagen original - histo_1.jpg')
plt.axis('off')
plt.show()

```

```python
# creamos las funciones para mostrar la imagen 2

img2 = io.imread('img/histo_2.jpg')

img2_norm = img2 / 255.0

plt.figure(figsize=(8,8))
plt.imshow(img2_norm)
plt.title('Imagen original - histo_2.jpg')
plt.axis('off')
plt.show()
```

#### 1) Realizar una transformación de color para convertir la imagen al espacio de color CMYK

```python
# Extraer la componente magenta de la imagen (que corresponde a la región tisular)
# Visualizar la imagen del canal magenta

import numpy as np
import matplotlib.pyplot as plt

def rgb_to_cmyk(img):

    # Separar canales RGB
    R = img[:,:,0]
    G = img[:,:,1]
    B = img[:,:,2]

    # Calcular componente negra (Key)
    K = 1 - np.maximum(np.maximum(R,G),B)
    K = np.clip(K,1e-8,1) # evitamos q divida por cero durante la conversión CMYK

    # Calcular componentes CMY
    C = (1-R-K)/(1-K)
    M = (1-G-K)/(1-K)
    Y = (1-B-K)/(1-K)

    return C,M,Y,K

# Convertir imagen RGB a CMYK
C1,M1,Y1,K1 = rgb_to_cmyk(img1_norm)

# Visualizar canal Magenta
# Este canal resalta mejor la región tisular
plt.figure(figsize=(6,6))
plt.imshow(M1,cmap='gray')
plt.title('Canal Magenta (histo_1)')
plt.axis('off')
plt.show()


```

```python
# Convertir RGB a CMYK para la foto 2
C2,M2,Y2,K2 = rgb_to_cmyk(img2_norm)

plt.figure(figsize=(6,6))
plt.imshow(M2,cmap='gray')
plt.title('Canal Magenta (histo_2)')
plt.axis('off')
plt.show()
```

#### 2) Umbralizar la imagen para separar los píxeles del fondo de la región tisular

```python
# Aplicar un filtro gaussiano de tamaño 5x5 y después utilizar el método de Otsu de manera que
# los píxeles correspondientes al lumen y al background de la imagen sean 1s y el resto de los píxeles tengan un valor de 0.
# Nota: Recordar que el método de Otsu requiere como input una imagen en el rango [0-255] en formato "uint8".
# Visualizar la máscara resultante

from skimage.filters import gaussian, threshold_otsu
from skimage import img_as_ubyte

# Aplicamos filtro Gaussiano sobre el canal Magenta obtenido de la conversión CMYK
suav = gaussian(M1, sigma=1.5) # el M1 seria el color magenta de la foto convertida / sigma es el control de la intensidad del desenfoque establecida en 1.5 para que elimine pequeñas variaciones y mantenga visible a los lumenes
                                # sin perder la estructura principal

# Convertir a uint8 para aplicar el filtro Otsu
suav_uint8 = img_as_ubyte(suav) # Convertir la imagen al formato uint8 (0-255),  requerido para aplicar correctamente el metodo de Otsu


# Obtener automáticamente el umbral óptimo
thresh = threshold_otsu(suav_uint8) # otsu nos ayuda a calcular automaticamente el umbral separando los tejidos del fondo 

print("Umbral Otsu =", thresh)

# Generar máscara binaria
# Blanco = tejido
# Negro = fondo y lúmenes
mascara_otsu1 = suav_uint8 >= thresh

# Visualizar máscara obtenida
plt.figure(figsize=(6,6))
plt.imshow(mascara_otsu1,cmap='gray')
plt.title('Mascara Otsu histo_1')
plt.axis('off')
plt.show()
```

```python
# aplicamos el mismo codigo para el histo_2

# aplicamos filtro gaussiano al canal magenta para reducir ruido
suav2 = gaussian(M2, sigma=1.5)

# convertir la imagen al formato uint8 rango 0 a 255 requerido por otsu
suav_uint82 = img_as_ubyte(suav2)

# obtener automaticamente el umbral que mejor separa el tejido del fondo
thresh2 = threshold_otsu(suav_uint82)

# mostrar el valor del umbral calculado
print("Umbral Otsu =", thresh2)

# se crea la mascara binaria usando el umbral obtenido
# blanco corresponde al tejido
# negro corresponde al fondo y lumenes
mascara_otsu2 = suav_uint82 >= thresh2

# visualizamos la mascara resultante
plt.figure(figsize=(6,6))
plt.imshow(mascara_otsu2,cmap='gray')
plt.title('Mascara Otsu histo_2')
plt.axis('off')
plt.show()
```

#### 3) Limpiar la imagen eliminando los artefactos de lumen (objetos blancos pequeños que no son lúmenes)

```python
# Utilizar la librería skimage.morphology.remove_small_objects para eliminar aquellos objetos cuya área sea menor a 300 píxeles
# Más información en https://scikit-image.org/docs/dev/api/skimage.morphology.html#skimage.morphology.remove_small_objects
# Visualizaer la máscara resultante


from skimage.morphology import remove_small_objects

# Elimina pequeñas regiones que corresponden a ruido en la imagen 
mascara_limpia1 = remove_small_objects(
    mascara_otsu1,
    min_size=300
)

# Visualizar el resultado
plt.figure(figsize=(6,6))
plt.imshow(mascara_limpia1,cmap='gray')
plt.title('Mascara limpia histo_1')
plt.axis('off')
plt.show()



```

```python
# elimina las regiones pequeñas del histo_2

mascara_limpia2 = remove_small_objects(
    mascara_otsu2,
    min_size=300
)

# muestra el resultado del histo_2
plt.figure(figsize=(6,6))
plt.imshow(mascara_limpia2,cmap='gray')
plt.title('Mascara limpia histo_2')
plt.axis('off')
plt.show()
```

#### 4) Rellenar con 0s el fondo de la imagen para quedarnos únicamente con los lúmenes

```python
# Aplicar el algoritmo de expansión a partir de semillas (region growing) de manera que únicamente los lúmenes sean blancos
# y el resto de la imagen negra. Pista: utilizar dos semillas. Nota: Se pueden fijar las semillas de manera manual, pero
# se valorará positivamente a aquell@s que desarrollen una función para encontrarlas automáticamente.
# Visualizar la máscara resultante. 

from collections import deque
import numpy as np
import matplotlib.pyplot as plt

def region_growing_fondo(mascara):

    # Obtiene las dimensiones de la mascara
    h, w = mascara.shape # h representa el numero de filas en la imagen , w representa al numero de columnas 

    # crea una mascara vacia donde se almacenara el fondo detectado
    fondo = np.zeros_like(mascara, dtype=bool)

    # Semillas ubicadas en las 4 esquinas de la imagen 
    semillas = [
        (0,0), # este valor representa a la esquina superior izquierda
        (h-1,0), # este valor representa a la esquina inferior izquierda
        (0,w-1), # este valor representa a la esquina superior derecha
        (h-1,w-1) # este valor representa a la esquina superior derecha
    ]

    cola = deque() # crea una cola para almacenar los pixeles q faltan por ser analizados durante la expancion 

    # Inicializar semillas pertenecientes al fondo
    for y,x in semillas: # recorre 1x1 las semillas definidas anteriormente

        if not mascara[y,x]: # hace una verificacion q la semilla pertenezca al fondo de la imagen

            fondo[y,x] = True # marca el pixel q a sido detectado como parte del fondo
            cola.append((y,x)) # agrega las semillas a la cola 

    # Vecindad de 4 conexiones, en otras palabras son las direcciones en q se van a mover las detecciones
    vecinos = [
        (-1,0), # se mueve una fila para arriba
        (1,0), # se mueve una fila para abajo
        (0,-1), # se mueve una columna para izquierda
        (0,1) # se mueve una columna para derecha
    ]

    # Expande la región sobre los píxeles negros conectados al borde
    while cola:

        y,x = cola.popleft() # obtiene el dato del siguiente pixel q va a ser analizado

        for dy,dx in vecinos: # revisa las posiciones adyacentes del pixel actual

            ny = y + dy
            nx = x + dx

            if 0 <= ny < h and 0 <= nx < w: # verifica las coordenadas q esten dentro de la imagen

                if (not mascara[ny,nx]) and (not fondo[ny,nx]): # hace comprobacion si el pixel pertenece al fondo o aun no a sido visitado

                    fondo[ny,nx] = True # marca al pixel como parte del fondo
                    cola.append((ny,nx)) # agrega el pixel a la cola y continua con la expansion

    # Conserva unicamente a los lumenes q ya han sido detectados
    lumenes = (~mascara) & (~fondo)

    return lumenes

# Ejecuta la Region Growing estableciendo las variables de la mascara limpia
lumenes1 = region_growing_fondo(mascara_limpia1)

# Visualizar lúmenes detectados
plt.figure(figsize=(8,8))
plt.imshow(lumenes1, cmap='gray')
plt.title('PUNTO 4 - Solo lúmenes histo_1')
plt.axis('off')
plt.show()
```

```python
# ejecuta la region growing para histo_2
lumenes2 = region_growing_fondo(mascara_limpia2)

# muestra los resultados
plt.figure(figsize=(8,8))
plt.imshow(lumenes2,cmap='gray')
plt.title('PUNTO 4 - Solo lúmenes histo_2')
plt.axis('off')
plt.show()
```

#### 5) Rellenar los objetos de los lúmenes

```python
# Rellenar los lúmenes con la función binary_fill_holes de la librería scipy.ndimage.morphology
# Visualizar la máscara resultante


# FUNCION PARA RELLENAR LOS HUECOS INTERNOS DE LOS LUMENES Y OBTENER REGIONES SOLIDAS PARA SU ANALISIS
from scipy.ndimage import binary_fill_holes

# Rellena posibles huecos internos en los lúmenes
mascara_final1 = binary_fill_holes(lumenes1)

plt.figure(figsize=(8,8))
plt.imshow(mascara_final1,cmap='gray')
plt.title('Lumenes rellenados histo_1')
plt.axis('off')
plt.show()

# FUNCION PARA ELIMINAR PEQUEÑOS ARTEFACTOS Y CONSERVAR UNICAMENTE LOS LUMENES DE MAYOR INTERES 
from skimage.morphology import remove_small_objects

# Elimina pequeños lumenes y artefactos residuales
mascara_final1 = remove_small_objects(
    mascara_final1,
    min_size=1000
)

plt.figure(figsize=(8,8))
plt.imshow(mascara_final1,cmap='gray')
plt.title('Lumenes filtrados histo_1')
plt.axis('off')
plt.show()

# FUNCION PARA IDENTIFICAR Y SELECCIONAR EL LUMEN PRINCIPAL (LUMEN DE MAYOR AREA)
from skimage.measure import label, regionprops

# Etiqueta los lumenes detectados
etiquetas = label(mascara_final1)

# Obtiene propiedades geometricas
props = regionprops(etiquetas)

# Selecciona el lumen de mayor area
prop_mayor = max(props, key=lambda p: p.area)

# Crea mascaras unicamente para el lumen principal
mascara_lumen_mayor = etiquetas == prop_mayor.label

plt.figure(figsize=(8,8))
plt.imshow(mascara_lumen_mayor,cmap='gray')
plt.title('Solo lumen principal histo_1')
plt.axis('off')
plt.show()


```

```python
# FUNCION PARA RELLENAR LOS HUECOS INTERNOS DE LOS LUMENES Y OBTENER REGIONES SOLIDAS PARA SU ANALISIS

mascara_final2 = binary_fill_holes(lumenes2)

plt.figure(figsize=(8,8))
plt.imshow(mascara_final2,cmap='gray')
plt.title('Lúmenes rellenados histo_2')
plt.axis('off')
plt.show()


# FUNCION PARA ELIMINAR PEQUEÑOS ARTEFACTOS Y CONSERVAR UNICAMENTE LOS LUMENES DE MAYOR INTERES

mascara_final2 = remove_small_objects(
    mascara_final2,
    min_size=1000
)

plt.figure(figsize=(8,8))
plt.imshow(mascara_final2,cmap='gray')
plt.title('Lúmenes filtrados histo_2')
plt.axis('off')
plt.show()


# FUNCION PARA IDENTIFICAR Y SELECCIONAR EL LUMEN PRINCIPAL (LUMEN DE MAYOR AREA)

etiquetas2 = label(mascara_final2)

props2 = regionprops(etiquetas2)

prop_mayor2 = max(props2, key=lambda p: p.area)

mascara_lumen_mayor2 = etiquetas2 == prop_mayor2.label

plt.figure(figsize=(8,8))
plt.imshow(mascara_lumen_mayor2,cmap='gray')
plt.title('Solo lumen principal histo_2')
plt.axis('off')
plt.show()
```

#### 6) Detectar y dibujar los contornos de los lúmenes sobre la imagen original

```python
# Dibujar los contornos de los lúmenes en color verde sobre la imagen original RGB. Nota: Utilizar los flags necesarios
# para que los contornos en verde sean perfectamente visibles. 
# Visualizar la imagen superpuesta

from skimage.measure import find_contours

# detecta las coordenadas del contorno del lumen principal
contornos = find_contours(
    mascara_lumen_mayor,
    0.5
)

# dibuja el contorno verde sobre la imagen original
plt.figure(figsize=(8,8))
plt.imshow(img1_norm)

# hace un recorrido por los contornos detectados y los va superponiendo sobre la imagen y colorea 
for contorno in contornos: 

    plt.plot(
        contorno[:,1],
        contorno[:,0],
        color='lime',
        linewidth=3
    )

# muestra el resultado final con el lumen marcado en verde
plt.title('Contorno del lumen principal histo_1')
plt.axis('off')
plt.show()
```

```python
# detecta las coordenadas del contorno el find_contours ayuda a obtener las coordenadas del borde del lumen y el 0.5 lo usa como referencia para localizar la transicion entre los pixeles negros y blancos de la mascara binaria
contornos2 = find_contours(
    mascara_lumen_mayor2,
    0.5
)

# dibuja el contorno verde en la imagen original
plt.figure(figsize=(8,8))
plt.imshow(img2_norm)

# hace un recorrido por los contornos detectados y los va superponiendo sobre la imagen y colorea
for contorno in contornos2:  

    plt.plot(
        contorno[:,1],
        contorno[:,0],
        color='lime',
        linewidth=3
    )

# muestra el resultado final con el contorno coloreado
plt.title('Contorno del lumen principal histo_2')
plt.axis('off')
plt.show()
```

#### 7) Identificar y cropear el lumen más grande

```python
# Determinar cuál es el lumen de mayor área y hacer un crop del mismo sobre la imagen original RGB.
# Visualizar el lumen cropeado.


import matplotlib.pyplot as plt

# Obtiene las coordenadas del Bounding Box, estos datos se encuentran en skimage.measure-bbox : tupla ahi establece estas coordenadas (min_row, min_col, max_row, max_col)
minr, minc, maxr, maxc = prop_mayor.bbox 
# minr = fila donde comienza el lumen
# minc = columna donde comienza el lumen
# maxr = fila donde termina el lumen
# maxc = columna donde termina el lumen

# Agrega un pequeño margen para mejorar la visualización
margen = 10

# Recorta el lumen principal sobre la imagen RGB original
crop1 = img1_norm[
    max(0,minr-margen):maxr+margen,
    max(0,minc-margen):maxc+margen
]

# Visualiza el resultado del recorte
plt.figure(figsize=(6,6))
plt.imshow(crop1)
plt.title('PUNTO 7 - Lumen más grande histo_1')
plt.axis('off')
plt.show()

```

```python
minr, minc, maxr, maxc = prop_mayor2.bbox

margen = 10

crop2 = img2_norm[
    max(0,minr-margen):maxr+margen,
    max(0,minc-margen):maxc+margen
]

plt.figure(figsize=(6,6))
plt.imshow(crop2)
plt.title('PUNTO 7 - Lumen más grande histo_2')
plt.axis('off')
plt.show()
```

#### 8) Extraer 13 características geométricas que permitan caracterizar el lumen recortado 

```python
# Calcular las siguientes características del crop del lumen de mayor área, redondeando su valor hasta el cuarto decimal.
# 1) Área
# 2) Área de la bounding box
# 3) Área convexa
# 4) Exentricidad
# 5) Diámetro equivalente
# 6) Extensión
# 7) Diámetro Feret
# 8) Longitud del eje mayor
# 9) Longitud del eje menor
# 10) Orientación
# 11) Perímetro
# 12) Solidez
# 13) Compacidad


# histo_1

import numpy as np

# Área del lumen / round nos sirve para redondear el area del lumen a 4 decimales segun lo solicitado en el enunciado
area = round(prop_mayor.area, 4)

# Área del Bounding Box
area_bbox = round(prop_mayor.area_bbox, 4)

# Área de la envolvente convexa
area_convexa = round(prop_mayor.area_convex, 4)

# Medida de alargamiento de la región
excentricidad = round(prop_mayor.eccentricity, 4)

# Diámetro del círculo equivalente
diametro_equivalente = round(
    prop_mayor.equivalent_diameter_area,
    4
)

# Relación entre área y Bounding Box
extension = round(prop_mayor.extent, 4)

# Máxima distancia entre dos puntos del contorno
diametro_feret = round(
    prop_mayor.feret_diameter_max,
    4
)

# Longitud del eje mayor de la elipse equivalente
eje_mayor = round(
    prop_mayor.axis_major_length,
    4
)

# Longitud del eje menor de la elipse equivalente
eje_menor = round(
    prop_mayor.axis_minor_length,
    4
)

# Ángulo de orientación del objeto
orientacion = round(
    prop_mayor.orientation,
    4
)

# Longitud total del contorno
perimetro = round(
    prop_mayor.perimeter,
    4
)

# Relación entre área y área convexa
solidez = round(
    prop_mayor.solidity,
    4
)

# Medida de circularidad del lumen / calcula la compacidad del lumen usando la formula (4 pi area) dividido para el perimetro al cuadrado y nos indica q tan cercana es la forma del lumen al circulo
compacidad = round(
    (4 * np.pi * area) / (perimetro ** 2), # np.pi representa el valor de pi y viene de la clase importada de numpy
    4
)

# Mostrar resultados
print("1 Área:", area)
print("2 Área Bounding Box:", area_bbox)
print("3 Área Convexa:", area_convexa)
print("4 Excentricidad:", excentricidad)
print("5 Diámetro Equivalente:", diametro_equivalente)
print("6 Extensión:", extension)
print("7 Diámetro Feret:", diametro_feret)
print("8 Longitud Eje Mayor:", eje_mayor)
print("9 Longitud Eje Menor:", eje_menor)
print("10 Orientación:", orientacion)
print("11 Perímetro:", perimetro)
print("12 Solidez:", solidez)
print("13 Compacidad:", compacidad)
```
