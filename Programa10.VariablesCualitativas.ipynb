```python
 # Creación aleatoria de datos
import numpy as np
import random
random.seed(42)
```

```python
# crea una nueva variable de lista vacia 
nivel_estudios = []
#creando o ingresando a mi lista varable secundaria  y creo 110 veces secuandaria
#va sumando es decir 172 mas 110 mas 145 y asi se imprime en cualquier orden
nivel_estudios.extend(['Secundaria']*110)
nivel_estudios.extend(['Bachillerato']*172)
nivel_estudios.extend(['Grado']*145)
nivel_estudios.extend(['Master']*52)
nivel_estudios.extend(['Doctorado']*21)
nivel_estudios = np.random.permutation(nivel_estudios)
```

```python
print(nivel_estudios)
```

```python
# DIAGRAMA DE BARRAS
import matplotlib.pyplot as plt
import numpy as np
estudios, frecuencia = np.unique(nivel_estudios, return_counts=True)
print(estudios, frecuencia)
```

```python
plt.ylabel('Frecuencia absoluta')
plt.xlabel('Nivel de estudios')
plt.title('Diagrama de barras')
plt.bar(estudios, frecuencia) # https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.bar.html
plt.show()
```

```python
 # DIAGRAMA DE SECTORES
colores = {'red', 'blue', 'orange', 'green', 'cyan'}
plt.pie(frecuencia, labels=estudios, colors=colores, autopct='%1.2f%%')
plt.title('Diagrama de sectores')
plt.show()
```

## Tarea 
Implementar con pandas esta informacion en un data frame

```python
import pandas as pd
df = pd.DataFrame(nivel_estudios, columns=['nivel_de_estudios'])
df
```

```python
# Contar la frecuencia absoluta de cada nivel de estudios
# .value_counts() cuenta las repeticiones y genera la frecuencia automáticamente
frecuencias = df['nivel_de_estudios'].value_counts()

# Crear el gráfico de barras directamente desde ese conteo
plt.figure(figsize=(8, 5))
frecuencias.plot(kind='bar', color='skyblue', edgecolor='black')

# Personalizar los ejes exactamente como los necesitas
plt.title('Diagrama de Barras "Nivel de estudios"', fontsize=14, fontweight='bold')
plt.xlabel('Nivel de Estudios', fontsize=12)
plt.ylabel('Frecuencia Absoluta', fontsize=12)

# Rotar las etiquetas del eje X para que se lean perfectamente de forma horizontal
plt.xticks(rotation=0)

# Añadir rejilla al fondo solo en el eje Y para facilitar la lectura de los valores
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Mostrar el gráfico limpio
plt.tight_layout()
plt.show()
```

```python
#Crear el diagrama de sectores
plt.figure(figsize=(6, 6))

# autopct='%1.1f%%' sirve para mostrar el porcentaje automáticamente dentro de cada sector
frecuencias.plot(kind='pie', autopct='%1.1f%%', startangle=90, cmap='Pastel1')

# 3. Personalización del gráfico
plt.title('Distribución Porcentual por Nivel de Estudios', fontsize=14, fontweight='bold')

# Quitamos la etiqueta automática vertical del eje para que se vea más limpio
plt.ylabel('')

# Mostrar el gráfico bien centrado
plt.tight_layout()
plt.show()
```
